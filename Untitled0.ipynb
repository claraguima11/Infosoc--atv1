## 1. Entendimento do problema

O problema é de classificação supervisionada de imagens em 3 classes: asphalt,
belgian_blocks e offroad. As principais dificuldades identificadas foram:

- **Desbalanceamento severo**: a classe belgian_blocks tem muito menos amostras,
  o que tende a fazer o modelo ignorá-la durante o treino.
- **Variações de iluminação**: imagens noturnas perdem textura, tornando asphalt
  escuro visualmente similar a offroad.
- **Similaridade visual entre classes**: belgian_blocks e asphalt são ambos
  superfícies pavimentadas, com diferença sutil de textura que pode se perder
  em baixa resolução ou iluminação ruim.


| Experimento              | Mudança principal                 | F1 macro | F1 belgian         | F1 asphalt | F1 offroad |
| ------------------------ | --------------------------------- | -------- | ------------------ | ---------- | ---------- |
| Exp 1 — baseline + pesos | ResNet18 congelada, class weights | 0.66     | 0.48 *(rec ~0.31)* | 0.87       | 0.64       |
| Exp 2 — ajuste de pesos  | Rebalanceamento dos pesos         | 0.65     | 0.48 *(rec ~0.34)* | 0.85       | 0.63       |
| Exp 3 — augmentation     | ColorJitter (iluminação)          | 0.70     | 0.53 *(rec ~0.41)* | 0.89       | 0.67       |
| Exp 4 — fine-tuning      | Descongelou layer4                | 0.82     | 0.62               | 0.96       | 0.87       |
| Exp final                | layer4 + fc + pesos + aug p=0.4   | **0.79** | **0.59**           | **0.94**   | **0.87**   |

```python
import torchvision.transforms as transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]
IMG_SIZE = 224
BATCH_SIZE = 32

# Transformações baseline (Experimento 0 e 1)
train_transforms_baseline = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD)
])

val_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD)
])

# Transformações ampliadas (Experimento 2)
train_transforms_augmented = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(p=0.2),
    transforms.RandomRotation(20),

    transforms.RandomApply([
        transforms.ColorJitter(brightness=0.7, contrast=0.5)
    ], p=0.4),


    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD)
])



print('Transformações definidas.')

# Ajuste os caminhos conforme a sua pasta no Drive
train_path = '/content/drive/MyDrive/Desafio_Voxar/dataset_processed/dataset_processed/train'
test_path = '/content/drive/MyDrive/Desafio_Voxar/dataset_processed/dataset_processed/test'

train_dataset = ImageFolder(root=train_path, transform=train_transforms_augmented)
test_dataset = ImageFolder(root=test_path, transform=val_transforms)

# O DataLoader entrega as imagens em "batches" (lotes), ex: 32 por vez
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

print(f"Dataset de treino carregado com {len(train_dataset)} imagens.")
print(f"Classes identificadas: {train_dataset.classes}")
```

```python
import torch
import torch.nn as nn
from torchvision import models

modelo = models.resnet18(weights='IMAGENET1K_V1')

for param in modelo.parameters():
    param.requires_grad = False

for param in modelo.layer4.parameters():
    param.requires_grad = True

for param in modelo.fc.parameters():
    param.requires_grad = True

num_ftrs = modelo.fc.in_features
modelo.fc = nn.Linear(num_ftrs, 3)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
modelo = modelo.to(device)

print("Modelo preparado para 3 classes e enviado para a GPU!")
```

```python
import torch.optim as optim

weights = torch.tensor([1.25, 20.0, 6.66]).to(device)

# Loss Function com pesos para combater o desbalanceamento
criterion = nn.CrossEntropyLoss(weight=weights)

# Otimizador: quem ajusta os parafusos do modelo
optimizer = optim.Adam(
    list(modelo.layer4.parameters()) + list(modelo.fc.parameters()),
    lr=1e-4
)
```

```python
epochs = 5

for epoch in range(epochs):
    modelo.train() # Coloca o modelo em modo de treino
    running_loss = 0.0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad() # Limpa o erro anterior

        outputs = modelo(inputs) # A IA tenta adivinhar
        loss = criterion(outputs, labels) # Calcula o erro (usando os pesos das classes)

        loss.backward() # Descobre como melhorar
        optimizer.step() # Ajusta o modelo

        running_loss += loss.item()

    print(f"Época {epoch+1}/{epochs} - Erro médio: {running_loss/len(train_loader):.4f}")

print("Treinamento concluído!")
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

modelo.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for inputs, labels in test_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        outputs = modelo(inputs)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

# Relatório completo
print(classification_report(all_labels, all_preds,
      target_names=train_dataset.classes))

# Matriz de confusão
cm = confusion_matrix(all_labels, all_preds)
sns.heatmap(cm, annot=True, fmt='d',
            xticklabels=train_dataset.classes,
            yticklabels=train_dataset.classes)
plt.ylabel('Real'); plt.xlabel('Predito')
plt.title('Matriz de Confusão'); plt.show()
```

```python
# Análise de erros por tipo de confusão
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
fig.suptitle("Erros: Belgian Blocks classificado como Asphalt", fontsize=13)

erros_belgian_asphalt = [
    i for i, (p, l) in enumerate(zip(all_preds, all_labels))
    if l == train_dataset.class_to_idx['belgian_blocks']
    and p == train_dataset.class_to_idx['asphalt']
]
for ax, idx in zip(axes.flat, erros_belgian_asphalt[:8]):
    img_path, _ = test_dataset.samples[idx]
    ax.imshow(Image.open(img_path))
    ax.set_title("Real: belgian → Pred: asphalt", fontsize=8)
    ax.axis('off')
plt.tight_layout(); plt.show()

# Segundo grid: asphalt escuro → offroad
fig2, axes2 = plt.subplots(2, 4, figsize=(16, 8))
fig2.suptitle("Erros: Asphalt classificado como Offroad", fontsize=13)
erros_asphalt_offroad = [
    i for i, (p, l) in enumerate(zip(all_preds, all_labels))
    if l == train_dataset.class_to_idx['asphalt']
    and p == train_dataset.class_to_idx['offroad']
]
for ax, idx in zip(axes2.flat, erros_asphalt_offroad[:8]):
    img_path, _ = test_dataset.samples[idx]
    ax.imshow(Image.open(img_path))
    ax.set_title("Real: asphalt → Pred: offroad", fontsize=8)
    ax.axis('off')
plt.tight_layout(); plt.show()
```

Em particular, o fato de ~96% dos erros de asphalt ocorrerem nessa condição (44/46) reforça que a iluminação é um fator crítico para a robustez do modelo.

## 5. Análise crítica

### Onde o modelo funciona bem
- Asphalt em boa iluminação: F1 0.94, poucos erros
- Offroad: F1 0.87, a textura de terra batida é distintiva

### Onde o modelo falha
- Belgian_blocks: F1 0.63. O principal padrão de erro é a confusão com asphalt,
  especialmente em imagens de baixa resolução ou com sombras que suavizam a textura
  das pedras. A confusão é assimétrica: o modelo frequentemente chama belgian de
  asphalt, mas raramente o contrário — sugerindo que asphalt se tornou uma classe
  "default" para superfícies pavimentadas, provavelmente pelo desbalanceamento.

### Limitações da solução
- O conjunto de teste foi usado para guiar decisões de hiperparâmetros (pesos,
  augmentation). Idealmente, um validation set separado deveria ter sido criado
  dentro do treino para isso. Os resultados finais podem estar levemente otimistas.
- 5 épocas de fine-tuning podem ser insuficientes para convergência total da layer4.

### O que eu faria com mais tempo
- Criar um validation split para monitorar overfitting durante o treino
- Testar RandomErasing como augmentation, simulando sombras e oclusões parciais
- Investigar se imagens noturnas de asphalt têm distribuição diferente no treino e
  no teste — isso poderia explicar a confusão com offroad
